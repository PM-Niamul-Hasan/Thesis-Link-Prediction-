```python

import pandas as pd
import numpy as np

def generate_source_target_dataset(
    num_samples=999,
    max_node_id=999,
    output_file="source_target.csv"
):
    # Generate random node IDs
    source_nodes = np.random.randint(1, max_node_id + 1, size=num_samples)
    target_nodes = np.random.randint(1, max_node_id + 1, size=num_samples)

    # Generate binary labels
    labels = np.random.randint(0, 2, size=num_samples)

    # Create DataFrame
    df = pd.DataFrame({
        'Source Node': source_nodes,
        'Target Node': target_nodes,
        'Label': labels
    })

    # Save to CSV
    df.to_csv(output_file, index=False)

    # Calculate number of unique nodes
    unique_nodes = set(source_nodes) | set(target_nodes)
    num_nodes = len(unique_nodes)

    print(f"Dataset generated: {output_file} with {num_samples} edges and {num_nodes} nodes.")

if __name__ == "__main__":
    generate_source_target_dataset()
```

```python

import pandas as pd
import numpy as np
import random

def generate_user_info_dataset(num_rows=999, output_file="user_info.csv"):
    # Node ID: sequential integers from 1 to num_rows
    node_ids = np.arange(1, num_rows + 1)

    # Age: random integer between 18 and 70
    ages = np.random.randint(18, 70, size=num_rows)

    # Gender: randomly choose 'Male', 'Female', 'Other'
    genders = np.random.choice(['Male', 'Female', 'Other'], size=num_rows, p=[0.45, 0.45, 0.10])

    # Location: randomly pick from a sample list of locations
    locations = np.random.choice(['Dhaka', 'Chittagong', 'Khulna', 'Rajshahi', 'Sylhet', 'Barisal'], size=num_rows)

    # Interests: randomly assign 1-3 interests from a list, combined as comma-separated strings
    interest_pool = ['Sports', 'Music', 'Travel', 'Reading', 'Cooking', 'Technology', 'Art', 'Fitness']
    interests = []
    for _ in range(num_rows):
        count = np.random.randint(1, 4)
        selected = random.sample(interest_pool, count)
        interests.append(', '.join(selected))

    # Activity Level: randomly choose from 'Low', 'Medium', 'High'
    activity_levels = np.random.choice(['Low', 'Medium', 'High'], size=num_rows, p=[0.3, 0.5, 0.2])

    # Create DataFrame
    df = pd.DataFrame({
        'Node ID': node_ids,
        'Age': ages,
        'Gender': genders,
        'Location': locations,
        'Interests': interests,
        'Activity Level': activity_levels
    })

    # Save to CSV
    df.to_csv(output_file, index=False)
    print(f"Dataset generated: user_info dataset with {num_rows} rows saved as '{output_file}'")

if __name__ == "__main__":
    generate_user_info_dataset()
```

```python

import random

def generate_facebook_topological_dataset(num_nodes=999, num_edges=999, output_file="facebook_topological.txt"):
    """
    Generate a fake Facebook topological list dataset.

    Args:
        num_nodes (int): Number of unique user nodes (IDs from 1 to num_nodes).
        num_edges (int): Number of edges (friend pairs) to create.
        output_file (str): Filename to w.rite the edge list.

    The dataset is an undirected graph node edge list with no self-loops or duplicate nodes edges.
    """
    edges = set()

    while len(edges) < num_edges:
        u = random.randint(1, num_nodes)
        v = random.randint(1, num_nodes)
        if u != v:
            edge = tuple(sorted((u, v)))  # sort so (u,v) == (v,u)
            edges.add(edge)

    with open(output_file, 'w') as f:
        for u, v in sorted(edges):
            f.write(f"{u} {v}\n")

    print(f"Dataset generated: {output_file} with {len(edges)} edges and {num_nodes} nodes.")

if __name__ == "__main__":
    generate_facebook_topological_dataset()
```

```python

import pandas as pd
import numpy as np
import networkx as nx
import random
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import accuracy_score, roc_auc_score, precision_score, recall_score, f1_score, roc_curve, confusion_matrix, classification_report, precision_recall_curve
from sklearn.ensemble import RandomForestClassifier, VotingClassifier, BaggingClassifier, AdaBoostClassifier, GradientBoostingClassifier, StackingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from imblearn.over_sampling import SMOTE
from xgboost import XGBClassifier
import os
import tensorflow as tf
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, LayerNormalization, MultiHeadAttention, Flatten, Reshape, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras import regularizers
from tabulate import tabulate
import time

# For Jupyter Notebooks
try:
    get_ipython()
    %matplotlib inline
except:
    pass

try:
    import plotly.graph_objects as go
except ImportError:
    import subprocess
    import sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "plotly"])
    import plotly.graph_objects as go

def seed_everything(seed=42):
    np.random.seed(seed)
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    tf.random.set_seed(seed)

seed_everything()

# ===================================================
# EDA functions
# ===================================================

def visualize_all_columns(df, dataset_name):
    """Visualize all columns in a dataset with appropriate plots"""
    print(f"\nVisualizing all columns for {dataset_name}:")

    for col in df.columns:
        # Numeric columns
        if pd.api.types.is_numeric_dtype(df[col]):
            # Histogram for numeric data
            plt.figure(figsize=(10, 5))
            sns.histplot(df[col], bins=30, kde=True)
            plt.title(f"{dataset_name} - {col} Distribution")
            plt.show()

            # Boxplot for numeric data
            plt.figure(figsize=(10, 3))
            sns.boxplot(x=df[col])
            plt.title(f"{dataset_name} - {col} Boxplot")
            plt.show()

        # Categorical columns
        else:
            # Countplot for categorical data
            if df[col].nunique() <= 20:  # Only plot if reasonable number of categories
                plt.figure(figsize=(10, 5))
                sns.countplot(data=df, x=col)
                plt.xticks(rotation=45)
                plt.title(f"{dataset_name} - {col} Distribution")
                plt.show()

def create_correlation_plots(df, dataset_name):
    """Create correlation plots for numeric columns"""
    numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

    if len(numeric_cols) > 1:
        # Correlation matrix
        plt.figure(figsize=(12, 8))
        corr = df[numeric_cols].corr()
        sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', center=0)
        plt.title(f"{dataset_name} - Correlation Matrix")
        plt.tight_layout()
        plt.show()

        # Pair plots for small number of columns
        if len(numeric_cols) <= 6:
            sns.pairplot(df[numeric_cols])
            plt.suptitle(f"{dataset_name} - Pair Plots", y=1.02)
            plt.show()

def create_combined_correlation(datasets):
    """Create one combined correlation plot from all datasets"""
    combined_features = []

    for name, df in datasets.items():
        numeric_cols = df.select_dtypes(include=['int64', 'float64'])
        if not numeric_cols.empty:
            prefixed_cols = numeric_cols.add_prefix(f"{name}_")
            combined_features.append(prefixed_cols)

    if combined_features:
        combined_df = pd.concat(combined_features, axis=1)
        plt.figure(figsize=(20, 16))
        sns.heatmap(combined_df.corr(), cmap='coolwarm', center=0,
                   annot=True, fmt='.2f', linewidths=.5)
        plt.title("Combined Correlation Matrix Across All Datasets")
        plt.tight_layout()
        plt.show()

def perform_dataset_eda(dataset_name, df):
    """Perform comprehensive EDA for a dataset"""
    print(f"\n{'='*50}")
    print(f"Analyzing {dataset_name} Dataset")
    print(f"{'='*50}")

    # Basic info
    print(f"\nShape: {df.shape}")
    print("\nFirst 5 rows:")
    print(df.head())

    # Descriptive statistics
    print("\nDescriptive Statistics:")
    print(df.describe())

    # Unique values
    print("\nUnique Values per Column:")
    for col in df.columns:
        print(f"{col}: {df[col].nunique()} unique values")

    # Visualize all columns
    visualize_all_columns(df, dataset_name)

    # Create correlation plots
    create_correlation_plots(df, dataset_name)

# ===================================================
# END OF EDA FUNCTIONS
# ===================================================

def generate_features(G, link_pairs, nodes_df=None, use_node_features=True):
    features = []
    attr_len = 0 if nodes_df is None else nodes_df.shape[1] - 1

    for u, v in link_pairs:
        if u in G and v in G:
            # Basic topological features
            cn = len(list(nx.common_neighbors(G, u, v)))
            jc = next(nx.jaccard_coefficient(G, [(u, v)]))[2]
            pa = next(nx.preferential_attachment(G, [(u, v)]))[2]
            aa = next(nx.adamic_adar_index(G, [(u, v)]))[2]
            ra = next(nx.resource_allocation_index(G, [(u, v)]))[2]
            deg_u, deg_v = G.degree(u), G.degree(v)
            cluster_u, cluster_v = nx.clustering(G, u), nx.clustering(G, v)

            # Advanced features
            try:
                shortest_path = nx.shortest_path_length(G, u, v)
            except nx.NetworkXNoPath:
                shortest_path = 0

            # Community-based features
            try:
                communities = nx.community.greedy_modularity_communities(G)
                same_community = any(u in c and v in c for c in communities)
            except:
                same_community = False
        else:
            cn = jc = pa = aa = ra = deg_u = deg_v = cluster_u = cluster_v = shortest_path = 0
            same_community = False

        feat_vector = [
            cn, jc, pa, aa, ra, deg_u, deg_v, cluster_u, cluster_v,
            shortest_path, int(same_community)
        ]

        if use_node_features and nodes_df is not None:
            feat_u = nodes_df.loc[nodes_df['id'] == u].drop(columns='id').values.flatten()
            feat_v = nodes_df.loc[nodes_df['id'] == v].drop(columns='id').values.flatten()
            if feat_u.size == 0:
                feat_u = np.zeros(attr_len)
            if feat_v.size == 0:
                feat_v = np.zeros(attr_len)
            feat_vector.extend(feat_u)
            feat_vector.extend(feat_v)

        features.append(feat_vector)
    return np.array(features)

def load_csv_data(use_node_features=True):
    nodes_df = pd.read_csv("user_info.csv")
    nodes_df.rename(columns=lambda x: x.strip(), inplace=True)
    nodes_df.drop_duplicates(inplace=True)
    if 'Node ID' in nodes_df.columns:
        nodes_df.rename(columns={'Node ID': 'id'}, inplace=True)

    for col in nodes_df.columns:
        if nodes_df[col].dtype == 'object' and col != 'id':
            mode_val = nodes_df[col].mode()
            if not mode_val.empty:
                nodes_df[col] = nodes_df[col].fillna(mode_val[0])
            nodes_df[col] = LabelEncoder().fit_transform(nodes_df[col].astype(str))
        elif col != 'id':
            median_val = nodes_df[col].median()
            nodes_df[col] = nodes_df[col].fillna(median_val)

    edges_df = pd.read_csv("user_interaction.csv")
    edges_df.rename(columns=lambda x: x.strip(), inplace=True)
    edges_df.drop_duplicates(inplace=True)
    if 'Source Node' in edges_df.columns and 'Target Node' in edges_df.columns:
        edges_df.rename(columns={'Source Node': 'source', 'Target Node': 'target'}, inplace=True)
    edges_df.dropna(subset=['source', 'target'], inplace=True)
    edges_df = edges_df.astype({'source': int, 'target': int})

    labels_df = pd.read_csv("source_target.csv")
    labels_df.rename(columns=lambda x: x.strip(), inplace=True)
    labels_df.drop_duplicates(inplace=True)
    if all(col in labels_df.columns for col in ['Source Node', 'Target Node', 'Label']):
        labels_df.rename(columns={'Source Node': 'source', 'Target Node': 'target', 'Label': 'label'}, inplace=True)
    labels_df.dropna(subset=['source', 'target', 'label'], inplace=True)
    labels_df = labels_df.astype({'source': int, 'target': int, 'label': int})

    G = nx.Graph()
    G.add_edges_from(edges_df[['source', 'target']].values)

    if not use_node_features:
        nodes_df = None

    X = generate_features(G, labels_df[['source', 'target']].values, nodes_df, use_node_features)
    y = labels_df['label'].values

    return X, y, G, nodes_df

def load_facebook_data(use_node_features=False):
    fb_edges = []
    with open("facebook_topological.txt", 'r') as f:
        for line in f:
            if line.startswith("#") or line.strip() == "":
                continue
            parts = line.strip().split()
            if len(parts) != 2:
                continue
            try:
                u, v = map(int, parts)
                fb_edges.append((u, v))
            except ValueError:
                continue

    G = nx.Graph()
    G.add_edges_from(fb_edges)

    pos_links = random.sample(fb_edges, min(25000, len(fb_edges)))
    nodes = list(G.nodes())
    neg_links = set()
    while len(neg_links) < len(pos_links):
        u, v = random.sample(nodes, 2)
        if not G.has_edge(u, v) and u != v:
            neg_links.add((u, v))
    neg_links = list(neg_links)

    links = pos_links + neg_links
    labels = [1]*len(pos_links) + [0]*len(neg_links)

    nodes_df = pd.DataFrame({'id': nodes}) if use_node_features else None

    X = generate_features(G, links, nodes_df, use_node_features)
    y = np.array(labels)
    return X, y, G, nodes_df

def load_combined_data(use_node_features=True):
    print("Loading CSV dataset...")
    X_csv, y_csv, G_csv, nodes_df_csv = load_csv_data(use_node_features)
    print("Loading Facebook dataset...")
    X_fb, y_fb, G_fb, nodes_df_fb = load_facebook_data(use_node_features)

    print(f"CSV dataset shape: {X_csv.shape}, labels: {y_csv.shape}")
    print(f"Facebook dataset shape: {X_fb.shape}, labels: {y_fb.shape}")

    # Create combined dataset
    if use_node_features and nodes_df_csv is not None and nodes_df_fb is not None:
        len_csv = X_csv.shape[1]
        len_fb = X_fb.shape[1]
        if len_csv > len_fb:
            diff = len_csv - len_fb
            X_fb = np.hstack([X_fb, np.zeros((X_fb.shape[0], diff))])
        elif len_fb > len_csv:
            diff = len_fb - len_csv
            X_csv = np.hstack([X_csv, np.zeros((X_csv.shape[0], diff))])

    X_combined = np.vstack([X_csv, X_fb])
    y_combined = np.concatenate([y_csv, y_fb])

    print(f"Combined dataset shape: {X_combined.shape}, labels: {y_combined.shape}")

    # Create combined graph
    G_combined = nx.disjoint_union(G_csv, G_fb)

    # Add prefix to nodes to distinguish datasets
    mapping_csv = {node: f"CSV_{node}" for node in G_csv.nodes()}
    mapping_fb = {node: f"FB_{node}" for node in G_fb.nodes()}
    G_combined = nx.relabel_nodes(G_combined, {**mapping_csv, **mapping_fb})

    return X_combined, y_combined, G_csv, G_fb, G_combined

def run_kfold_cv(model, model_name, X, y, n_splits=5):
    """Perform K-Fold Cross Validation for a model"""
    kfold = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
    scaler = StandardScaler()
    smote = SMOTE(random_state=42)

    metrics = {
        'AUC': [],
        'Accuracy': [],
        'Precision': [],
        'Recall': [],
        'F1': [],
        'Fit Time': []
    }

    for fold, (train_idx, val_idx) in enumerate(kfold.split(X, y)):
        X_train, X_val = X[train_idx], X[val_idx]
        y_train, y_val = y[train_idx], y[val_idx]

        # Scale data
        X_train_scaled = scaler.fit_transform(X_train)
        X_val_scaled = scaler.transform(X_val)

        # Apply SMOTE
        X_train_res, y_train_res = smote.fit_resample(X_train_scaled, y_train)

        # Train model
        start_time = time.time()

        if model_name in ['KerasDNN', 'Transformer']:
            early_stop = EarlyStopping(
                monitor='val_auc',
                patience=5,
                mode='max',
                restore_best_weights=True,
                verbose=0
            )
            model.fit(
                X_train_res, y_train_res,
                epochs=100,
                batch_size=256,
                validation_data=(X_val_scaled, y_val),
                callbacks=[early_stop],
                verbose=0
            )
        else:
            model.fit(X_train_res, y_train_res)

        fit_time = time.time() - start_time

        # Predict
        if model_name in ['KerasDNN', 'Transformer']:
            probs = model.predict(X_val_scaled).flatten()
            preds = (probs > 0.5).astype(int)
        else:
            probs = model.predict_proba(X_val_scaled)[:, 1]
            preds = model.predict(X_val_scaled)

        # Calculate metrics
        metrics['AUC'].append(roc_auc_score(y_val, probs))
        metrics['Accuracy'].append(accuracy_score(y_val, preds))
        metrics['Precision'].append(precision_score(y_val, preds))
        metrics['Recall'].append(recall_score(y_val, preds))
        metrics['F1'].append(f1_score(y_val, preds))
        metrics['Fit Time'].append(fit_time)

        print(f"  Fold {fold+1}/{n_splits} - AUC: {metrics['AUC'][-1]:.4f}, F1: {metrics['F1'][-1]:.4f}, Time: {fit_time:.2f}s")

    return metrics

def evaluate_kfold_performance(model, model_name, X, y):
    """Evaluate model with different K-Fold configurations and visualize results"""
    kfold_results = {}
    fold_configs = [3, 4, 7]

    print(f"\n{'='*50}")
    print(f"K-Fold Cross Validation for {model_name}")
    print(f"{'='*50}")

    for n_splits in fold_configs:
        print(f"\nRunning {n_splits}-Fold Cross Validation...")
        kfold_results[n_splits] = run_kfold_cv(model, model_name, X, y, n_splits)

    # Create visualizations
    plt.figure(figsize=(14, 10))

    # Plot AUC for different folds
    plt.subplot(2, 2, 1)
    for n_splits, res in kfold_results.items():
        plt.plot(range(1, n_splits+1), res['AUC'], 'o-', label=f'{n_splits}-Fold')
    plt.title('AUC by Fold Configuration')
    plt.xlabel('Fold')
    plt.ylabel('AUC')
    plt.legend()
    plt.grid(True)

    # Plot F1 for different folds
    plt.subplot(2, 2, 2)
    for n_splits, res in kfold_results.items():
        plt.plot(range(1, n_splits+1), res['F1'], 'o-', label=f'{n_splits}-Fold')
    plt.title('F1 Score by Fold Configuration')
    plt.xlabel('Fold')
    plt.ylabel('F1 Score')
    plt.legend()
    plt.grid(True)

    # Plot average metrics by fold config
    plt.subplot(2, 2, 3)
    avg_metrics = {
        'AUC': [np.mean(res['AUC']) for res in kfold_results.values()],
        'F1': [np.mean(res['F1']) for res in kfold_results.values()],
        'Accuracy': [np.mean(res['Accuracy']) for res in kfold_results.values()],
    }
    x = np.arange(len(fold_configs))
    width = 0.25

    plt.bar(x - width, avg_metrics['AUC'], width, label='AUC')
    plt.bar(x, avg_metrics['F1'], width, label='F1')
    plt.bar(x + width, avg_metrics['Accuracy'], width, label='Accuracy')

    plt.title('Average Metrics by Fold Configuration')
    plt.xlabel('Fold Configuration')
    plt.ylabel('Score')
    plt.xticks(x, fold_configs)
    plt.legend()
    plt.grid(True, axis='y')

    # Plot training time
    plt.subplot(2, 2, 4)
    avg_times = [np.mean(res['Fit Time']) for res in kfold_results.values()]
    plt.bar(fold_configs, avg_times, color='purple')
    plt.title('Average Training Time by Fold Configuration')
    plt.xlabel('Number of Folds')
    plt.ylabel('Time (seconds)')
    plt.grid(True, axis='y')

    plt.tight_layout()
    plt.show()

    # Create detailed results table
    print("\nK-Fold Cross Validation Summary:")
    table_data = []
    for n_splits, res in kfold_results.items():
        row = [f"{n_splits}-Fold"]
        for metric in ['AUC', 'Accuracy', 'Precision', 'Recall', 'F1']:
            mean_val = np.mean(res[metric])
            std_val = np.std(res[metric])
            row.append(f"{mean_val:.4f} ± {std_val:.4f}")
        row.append(f"{np.mean(res['Fit Time']):.2f}s")
        table_data.append(row)

    headers = ["Fold Config", "AUC", "Accuracy", "Precision", "Recall", "F1", "Avg Time"]
    print(tabulate(table_data, headers=headers, tablefmt="grid", floatfmt=".4f"))

    return kfold_results

def train_various_ensembles(X, y):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled, y, test_size=0.25, stratify=y, random_state=42
    )

    smote = SMOTE(random_state=42)
    X_train, y_train = smote.fit_resample(X_train, y_train)

    # Enhanced base models
    rf = RandomForestClassifier(
        n_estimators=200,
        max_depth=12,
        min_samples_split=5,
        min_samples_leaf=2,
        max_features='sqrt',
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    )

    svm = SVC(
        probability=True,
        kernel='rbf',
        C=1.0,
        gamma='scale',
        class_weight='balanced',
        random_state=42
    )

    knn = KNeighborsClassifier(
        n_neighbors=5,
        weights='distance',
        algorithm='auto',
        leaf_size=30,
        n_jobs=-1
    )

    mlp = MLPClassifier(
        hidden_layer_sizes=(256, 128, 64),
        activation='relu',
        solver='adam',
        alpha=0.0001,
        batch_size=256,
        learning_rate='adaptive',
        early_stopping=True,
        validation_fraction=0.1,
        n_iter_no_change=10,
        max_iter=500,
        random_state=42
    )

    xgb = XGBClassifier(
        n_estimators=200,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective='binary:logistic',
        n_jobs=-1,
        random_state=42,
        eval_metric='auc'
    )

    models = [
        ('RandomForest', rf),
        ('SVM', svm),
        ('k-NN', knn),
        ('MLP', mlp),
        ('XGBoost', xgb)
    ]

    results = {}
    kfold_results = {}

    for name, model in models:
        print(f"\nTraining {name}...")

        # Run K-Fold validation first
        kfold_results[name] = evaluate_kfold_performance(model, name, X, y)

        # Now train on full data
        model.fit(X_train, y_train)

        probs = model.predict_proba(X_test)[:, 1]
        preds = model.predict(X_test)

        results[name] = {
            'model': model,
            'auc': roc_auc_score(y_test, probs),
            'accuracy': accuracy_score(y_test, preds),
            'precision': precision_score(y_test, preds),
            'recall': recall_score(y_test, preds),
            'f1': f1_score(y_test, preds),
            'y_pred': preds,
            'probs': probs
        }
        print(f"{name} trained successfully. AUC: {results[name]['auc']:.4f}")

    # Enhanced Keras DNN
    print("\nTraining Enhanced Keras DNN...")
    keras_dnn = Sequential([
        Input(shape=(X_train.shape[1],)),
        Dense(256, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
        BatchNormalization(),
        Dropout(0.4),
        Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
        BatchNormalization(),
        Dropout(0.3),
        Dense(64, activation='relu'),
        Dropout(0.2),
        Dense(1, activation='sigmoid')
    ])

    keras_dnn.compile(
        optimizer=Adam(learning_rate=0.0005),
        loss='binary_crossentropy',
        metrics=['accuracy', tf.keras.metrics.AUC()]
    )

    # Run K-Fold validation first
    kfold_results['KerasDNN'] = evaluate_kfold_performance(keras_dnn, 'KerasDNN', X, y)

    # Now train on full data
    early_stop = EarlyStopping(
        monitor='val_auc',
        patience=10,
        mode='max',
        restore_best_weights=True
    )

    keras_dnn.fit(
        X_train, y_train,
        epochs=200,
        batch_size=128,
        validation_split=0.15,
        callbacks=[early_stop],
        verbose=0
    )

    keras_probs = keras_dnn.predict(X_test).flatten()
    keras_preds = (keras_probs > 0.5).astype(int)

    results['KerasDNN'] = {
        'model': keras_dnn,
        'auc': roc_auc_score(y_test, keras_probs),
        'accuracy': accuracy_score(y_test, keras_preds),
        'precision': precision_score(y_test, keras_preds),
        'recall': recall_score(y_test, keras_preds),
        'f1': f1_score(y_test, keras_preds),
        'y_pred': keras_preds,
        'probs': keras_probs
    }
    print(f"Keras DNN trained successfully. AUC: {results['KerasDNN']['auc']:.4f}")

    # Enhanced Transformer Model
    print("\nTraining Enhanced Transformer Model...")
    def transformer_model(input_dim, num_heads=8, ff_dim=128, dropout=0.2):
        inputs = Input(shape=(input_dim,))
        x = Reshape((1, input_dim))(inputs)
        x = LayerNormalization(epsilon=1e-6)(x)

        attn_output = MultiHeadAttention(
            num_heads=num_heads,
            key_dim=input_dim//num_heads
        )(x, x)
        attn_output = Dropout(dropout)(attn_output)
        x = LayerNormalization(epsilon=1e-6)(x + attn_output)

        ff_output = Sequential([
            Dense(ff_dim, activation="gelu"),
            Dropout(dropout),
            Dense(input_dim)
        ])(x)
        ff_output = Dropout(dropout)(ff_output)
        x = LayerNormalization(epsilon=1e-6)(x + ff_output)

        x = Flatten()(x)
        x = Dense(128, activation="relu")(x)
        x = Dropout(0.3)(x)
        x = Dense(64, activation="relu")(x)
        outputs = Dense(1, activation="sigmoid")(x)

        model = Model(inputs=inputs, outputs=outputs)
        model.compile(
            optimizer=Adam(learning_rate=0.0001),
            loss="binary_crossentropy",
            metrics=["accuracy", tf.keras.metrics.AUC()]
        )
        return model

    transformer = transformer_model(X_train.shape[1])

    # Run K-Fold validation first
    kfold_results['Transformer'] = evaluate_kfold_performance(transformer, 'Transformer', X, y)

    # Now train on full data
    transformer.fit(
        X_train, y_train,
        epochs=150,
        batch_size=128,
        validation_split=0.15,
        callbacks=[early_stop],
        verbose=0
    )

    transformer_probs = transformer.predict(X_test).flatten()
    transformer_preds = (transformer_probs > 0.5).astype(int)

    results['Transformer'] = {
        'model': transformer,
        'auc': roc_auc_score(y_test, transformer_probs),
        'accuracy': accuracy_score(y_test, transformer_preds),
        'precision': precision_score(y_test, transformer_preds),
        'recall': recall_score(y_test, transformer_preds),
        'f1': f1_score(y_test, transformer_preds),
        'y_pred': transformer_preds,
        'probs': transformer_probs
    }
    print(f"Transformer trained successfully. AUC: {results['Transformer']['auc']:.4f}")

    # Ensembles
    print("\nTraining Enhanced Voting Ensemble...")
    voting = VotingClassifier(
        estimators=[
            ('rf', rf),
            ('xgb', xgb),
            ('mlp', mlp)
        ],
        voting='soft',
        n_jobs=-1
    )

    # Run K-Fold validation first
    kfold_results['VotingEnsemble'] = evaluate_kfold_performance(voting, 'VotingEnsemble', X, y)

    # Now train on full data
    voting.fit(X_train, y_train)
    voting_probs = voting.predict_proba(X_test)[:, 1]
    voting_preds = voting.predict(X_test)

    results['VotingEnsemble'] = {
        'model': voting,
        'auc': roc_auc_score(y_test, voting_probs),
        'accuracy': accuracy_score(y_test, voting_preds),
        'precision': precision_score(y_test, voting_preds),
        'recall': recall_score(y_test, voting_preds),
        'f1': f1_score(y_test, voting_preds),
        'y_pred': voting_preds,
        'probs': voting_probs
    }
    print(f"Voting Ensemble trained successfully. AUC: {results['VotingEnsemble']['auc']:.4f}")

    print("\nTraining Stacking Ensemble...")
    stack = StackingClassifier(
        estimators=[
            ('rf', rf),
            ('xgb', xgb),
            ('svm', svm)
        ],
        final_estimator=LogisticRegression(
            penalty='l2',
            C=1.0,
            class_weight='balanced',
            max_iter=1000,
            random_state=42,
            n_jobs=-1
        ),
        n_jobs=-1
    )

    # Run K-Fold validation first
    kfold_results['StackingEnsemble'] = evaluate_kfold_performance(stack, 'StackingEnsemble', X, y)

    # Now train on full data
    stack.fit(X_train, y_train)
    stack_probs = stack.predict_proba(X_test)[:, 1]
    stack_preds = stack.predict(X_test)

    results['StackingEnsemble'] = {
        'model': stack,
        'auc': roc_auc_score(y_test, stack_probs),
        'accuracy': accuracy_score(y_test, stack_preds),
        'precision': precision_score(y_test, stack_preds),
        'recall': recall_score(y_test, stack_preds),
        'f1': f1_score(y_test, stack_preds),
        'y_pred': stack_preds,
        'probs': stack_probs
    }
    print(f"Stacking Ensemble trained successfully. AUC: {results['StackingEnsemble']['auc']:.4f}")

    print("\nTraining Bagging Ensemble...")
    bagging = BaggingClassifier(
        estimator=RandomForestClassifier(
            n_estimators=100,
            max_depth=10,
            min_samples_split=5,
            min_samples_leaf=2,
            class_weight='balanced',
            random_state=42,
            n_jobs=-1
        ),
        n_estimators=10,
        max_samples=0.8,
        max_features=0.8,
        random_state=42,
        n_jobs=-1
    )

    # Run K-Fold validation first
    kfold_results['BaggingEnsemble'] = evaluate_kfold_performance(bagging, 'BaggingEnsemble', X, y)

    # Now train on full data
    bagging.fit(X_train, y_train)
    bagging_probs = bagging.predict_proba(X_test)[:, 1]
    bagging_preds = bagging.predict(X_test)

    results['BaggingEnsemble'] = {
        'model': bagging,
        'auc': roc_auc_score(y_test, bagging_probs),
        'accuracy': accuracy_score(y_test, bagging_preds),
        'precision': precision_score(y_test, bagging_preds),
        'recall': recall_score(y_test, bagging_preds),
        'f1': f1_score(y_test, bagging_preds),
        'y_pred': bagging_preds,
        'probs': bagging_probs
    }
    print(f"Bagging Ensemble trained successfully. AUC: {results['BaggingEnsemble']['auc']:.4f}")

    print("\nTraining AdaBoost Ensemble...")
    adaboost = AdaBoostClassifier(
        estimator=DecisionTreeClassifier(
            max_depth=3,
            min_samples_split=5,
            min_samples_leaf=2,
            class_weight='balanced',
            random_state=42
        ),
        n_estimators=100,
        learning_rate=0.1,
        random_state=42
    )

    # Run K-Fold validation first
    kfold_results['AdaBoostEnsemble'] = evaluate_kfold_performance(adaboost, 'AdaBoostEnsemble', X, y)

    # Now train on full data
    adaboost.fit(X_train, y_train)
    adaboost_probs = adaboost.predict_proba(X_test)[:, 1]
    adaboost_preds = adaboost.predict(X_test)

    results['AdaBoostEnsemble'] = {
        'model': adaboost,
        'auc': roc_auc_score(y_test, adaboost_probs),
        'accuracy': accuracy_score(y_test, adaboost_preds),
        'precision': precision_score(y_test, adaboost_preds),
        'recall': recall_score(y_test, adaboost_preds),
        'f1': f1_score(y_test, adaboost_preds),
        'y_pred': adaboost_preds,
        'probs': adaboost_probs
    }
    print(f"AdaBoost Ensemble trained successfully. AUC: {results['AdaBoostEnsemble']['auc']:.4f}")

    print("\nTraining Gradient Boosting Ensemble...")
    gbm = GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42
    )

    # Run K-Fold validation first
    kfold_results['GradientBoostingEnsemble'] = evaluate_kfold_performance(gbm, 'GradientBoostingEnsemble', X, y)

    # Now train on full data
    gbm.fit(X_train, y_train)
    gbm_probs = gbm.predict_proba(X_test)[:, 1]
    gbm_preds = gbm.predict(X_test)

    results['GradientBoostingEnsemble'] = {
        'model': gbm,
        'auc': roc_auc_score(y_test, gbm_probs),
        'accuracy': accuracy_score(y_test, gbm_preds),
        'precision': precision_score(y_test, gbm_preds),
        'recall': recall_score(y_test, gbm_preds),
        'f1': f1_score(y_test, gbm_preds),
        'y_pred': gbm_preds,
        'probs': gbm_probs
    }
    print(f"Gradient Boosting Ensemble trained successfully. AUC: {results['GradientBoostingEnsemble']['auc']:.4f}")

    return results, kfold_results, X_train, X_test, y_train, y_test

def plot_graphs_separately(results, y_test, best_model_name, best_res):
    # 1. ROC Curve (separate figure)
    plt.figure(figsize=(8, 6))
    for name, res in results.items():
        fpr, tpr, _ = roc_curve(y_test, res['probs'])
        plt.plot(fpr, tpr, lw=2, label=f"{name} (AUC={res['auc']:.3f})")
    plt.plot([0, 1], [0, 1], 'k--')
    plt.title("ROC Curves")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend(loc='lower right')
    plt.grid(True)
    plt.show()

    # 2. Confusion Matrix (separate figure)
    plt.figure(figsize=(6, 6))
    cm = confusion_matrix(y_test, best_res['y_pred'])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f"Confusion Matrix - {best_model_name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

    # 3. Precision-Recall Curve (separate figure)
    plt.figure(figsize=(8, 6))
    precision, recall, _ = precision_recall_curve(y_test, best_res['probs'])
    plt.plot(recall, precision, lw=2, label=f"{best_model_name}")
    plt.title("Precision-Recall Curve")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.grid(True)
    plt.legend()
    plt.show()

    # Print classification report
    print(f"\nClassification Report for {best_model_name}:")
    print(classification_report(y_test, best_res['y_pred']))

def plot_interactive_network(G, title="Interactive Network Graph"):
    # Create positions for all nodes
    pos = nx.spring_layout(G, seed=42)

    # Create edge traces
    edge_x = []
    edge_y = []
    for edge in G.edges():
        x0, y0 = pos[edge[0]]
        x1, y1 = pos[edge[1]]
        edge_x.extend([x0, x1, None])
        edge_y.extend([y0, y1, None])

    edge_trace = go.Scatter(
        x=edge_x, y=edge_y,
        line=dict(width=0.5, color='#888'),
        hoverinfo='none',
        mode='lines')

    # Create node traces
    node_x = []
    node_y = []
    node_text = []
    node_adjacencies = []
    for node in G.nodes():
        x, y = pos[node]
        node_x.append(x)
        node_y.append(y)
        neighbors = list(G.neighbors(node))
        node_adjacencies.append(len(neighbors))
        node_text.append(f'Node: {node}<br># Connections: {len(neighbors)}')

    node_trace = go.Scatter(
        x=node_x, y=node_y,
        mode='markers',
        hoverinfo='text',
        text=node_text,
        marker=dict(
            showscale=True,
            colorscale='YlGnBu',
            reversescale=True,
            color=node_adjacencies,
            size=10,
            colorbar=dict(
                thickness=15,
                title='Node Connections',
                xanchor='left',
                titleside='right'
            ),
            line_width=2)
    )

    # Create figure
    fig = go.Figure(data=[edge_trace, node_trace],
                 layout=go.Layout(
                    title=f'<b>{title}</b>',
                    titlefont_size=16,
                    showlegend=False,
                    hovermode='closest',
                    margin=dict(b=20,l=5,r=5,t=40),
                    xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
                    yaxis=dict(showgrid=False, zeroline=False, showticklabels=False))
                )

    # Show the figure
    fig.show()

def generate_feature_names(nodes_df_csv, use_node_features=True):
    """Generate meaningful feature names for SHAP analysis"""
    base_names = [
        'common_neighbors',
        'jaccard_coefficient',
        'preferential_attachment',
        'adamic_adar',
        'resource_allocation',
        'degree_u',
        'degree_v',
        'clustering_u',
        'clustering_v',
        'shortest_path',
        'same_community'
    ]

    if use_node_features and nodes_df_csv is not None:
        node_feature_names = nodes_df_csv.drop(columns='id', errors='ignore').columns.tolist()
        u_names = [f"u_{col}" for col in node_feature_names]
        v_names = [f"v_{col}" for col in node_feature_names]
        feature_names = base_names + u_names + v_names
    else:
        feature_names = base_names

    return feature_names

def perform_shap_analysis(model, model_name, X_train, X_test, feature_names):
    """Perform SHAP analysis for model interpretability with Keras support"""
    print(f"\n{'='*50}")
    print(f"SHAP Analysis for {model_name}")
    print(f"{'='*50}")

    try:
        import shap
    except ImportError:
        import subprocess
        import sys
        subprocess.check_call([sys.executable, "-m", "pip", "install", "shap"])
        import shap

    try:
        # Select a representative sample for SHAP
        background = shap.utils.sample(X_train, 100)
        test_sample = X_test[:100]  # Use first 100 test samples

        # Handle different model types
        if model_name in ['RandomForest', 'XGBoost', 'BaggingEnsemble',
                          'AdaBoostEnsemble', 'GradientBoostingEnsemble']:
            print("Using TreeExplainer")
            explainer = shap.TreeExplainer(model)
            shap_values = explainer.shap_values(test_sample)

            # Handle binary classification output
            if isinstance(shap_values, list) and len(shap_values) == 2:
                shap_values = shap_values[1]  # Use SHAP values for positive class

        elif model_name in ['KerasDNN', 'Transformer']:
            print("Using GradientExplainer for deep learning model")

            # Create prediction function
            def predict_fn(data):
                return model.predict(data).flatten()

            explainer = shap.KernelExplainer(predict_fn, background)
            shap_values = explainer.shap_values(test_sample, nsamples=50)

            # Handle binary classification output
            if isinstance(shap_values, list) and len(shap_values) == 2:
                shap_values = shap_values[1]

        else:  # For other models (SVM, MLP, etc.)
            print("Using KernelExplainer")

            def predict_fn(data):
                return model.predict_proba(data)[:, 1]

            explainer = shap.KernelExplainer(predict_fn, background)
            shap_values = explainer.shap_values(test_sample, nsamples=100)

        # ===================================================
        # Visualization Section
        # ===================================================

        # 1. Summary Plot (Feature Importance)
        plt.figure(figsize=(12, 8))
        shap.summary_plot(shap_values, test_sample, feature_names=feature_names, show=False)
        plt.title(f"{model_name} - Global Feature Importance")
        plt.tight_layout()
        plt.show()

        # 2. Bar Plot (Mean |SHAP Value|)
        plt.figure(figsize=(12, 8))
        shap.summary_plot(shap_values, test_sample, feature_names=feature_names,
                          plot_type="bar", show=False)
        plt.title(f"{model_name} - Mean |SHAP Value|")
        plt.tight_layout()
        plt.show()

        # 3. Local Explanations
        print("\nLocal Explanations for Selected Instances:")
        for i in range(min(3, test_sample.shape[0])):
            plt.figure(figsize=(12, 6))
            shap.force_plot(
                explainer.expected_value,
                shap_values[i],
                test_sample[i],
                feature_names=feature_names,
                matplotlib=True,
                show=False
            )
            plt.title(f"{model_name} - SHAP Explanation for Instance {i+1}")
            plt.tight_layout()
            plt.show()

        # 4. Dependence Plots
        mean_abs_shap = np.abs(shap_values).mean(axis=0)
        top_indices = np.argsort(mean_abs_shap)[-3:][::-1]

        for idx in top_indices:
            plt.figure(figsize=(10, 6))
            shap.dependence_plot(
                idx,
                shap_values,
                test_sample,
                feature_names=feature_names,
                show=False
            )
            plt.title(f"{model_name} - Dependence Plot for {feature_names[idx]}")
            plt.tight_layout()
            plt.show()

    except Exception as e:
        print(f"Error in SHAP analysis for {model_name}: {str(e)}")
        import traceback
        traceback.print_exc()

def main():
    print("Running enhanced combined dataset training & evaluation...\n")

    # ===================================================
    # Load datasets for EDA
    # ===================================================
    datasets = {
        "UserInfo": pd.read_csv("user_info.csv"),
        "Interaction": pd.read_csv("user_interaction.csv"),
        "SourceTarget": pd.read_csv("source_target.csv"),
        "Facebook": pd.read_csv("facebook_topological.txt", sep=" ", header=None, names=['source', 'target'])
    }

    # Perform EDA for each dataset
    for name, df in datasets.items():
        perform_dataset_eda(name, df)

    # Create combined correlation plot
    create_combined_correlation(datasets)

    # ===================================================
    # Load combined dataset
    # ===================================================
    use_node_features = True
    X, y, G_csv, G_fb, G_combined = load_combined_data(use_node_features)

    # ===================================================
    # Save and display combined dataset
    # ===================================================
    # Get nodes_df for feature name generation
    nodes_df_csv = pd.read_csv("user_info.csv")
    if 'Node ID' in nodes_df_csv.columns:
        nodes_df_csv.rename(columns={'Node ID': 'id'}, inplace=True)

    feature_names = generate_feature_names(nodes_df_csv, use_node_features)

    # Create combined dataset DataFrame
    combined_df = pd.DataFrame(X, columns=feature_names)
    combined_df['label'] = y

    # Save to file
    combined_df.to_csv("combined_dataset.csv", index=False)
    print("\nSaved combined dataset to 'combined_dataset.csv'")

    # Print combined dataset info
    print("\nCombined Dataset Information:")
    print(f"Shape: {combined_df.shape}")
    print("\nFirst 5 rows:")
    print(combined_df.head())
    print("\nDescriptive Statistics:")
    print(combined_df.describe())
    print("\nLabel Distribution:")
    print(combined_df['label'].value_counts())

    # Add to datasets for EDA
    datasets["Combined"] = combined_df
    perform_dataset_eda("Combined", combined_df)

    # ===================================================
    # Train models
    # ===================================================
    results, kfold_results, X_train, X_test, y_train, y_test = train_various_ensembles(X, y)

    best_model_name = max(results, key=lambda k: results[k]['auc'])
    best_res = results[best_model_name]

    print("\n=== Comprehensive Model Performance ===")
    print(f"{'Model':25} | {'AUC':<6} | {'Accuracy':<8} | {'Precision':<8} | {'Recall':<6} | {'F1':<6}")
    print("-" * 80)
    for name, res in results.items():
        print(f"{name:25} | {res['auc']:.4f} | {res['accuracy']:.4f}  | {res['precision']:.4f}  | {res['recall']:.4f} | {res['f1']:.4f}")

    print("\n=== Best Model ===")
    print(f"Model: {best_model_name}")
    print(f"AUC: {best_res['auc']:.4f}")
    print(f"Accuracy: {best_res['accuracy']:.4f}")
    print(f"Precision: {best_res['precision']:.4f}")
    print(f"Recall: {best_res['recall']:.4f}")
    print(f"F1 Score: {best_res['f1']:.4f}")

    # Plot the graphs
    plot_graphs_separately(results, y_test, best_model_name, best_res)

    print("\nVisualizing network graphs...")
    plot_interactive_network(G_csv, title="CSV Dataset Network Graph")
    plot_interactive_network(G_fb, title="Facebook Dataset Network Graph")
    plot_interactive_network(G_combined, title="Combined Network Graph")

    # ===================================================
    # XAI Analysis for Best Model
    # ===================================================
    print("\nStarting XAI Analysis with SHAP for Best Model...")
    print(f"\nAnalyzing best model: {best_model_name}")
    perform_shap_analysis(
        best_res['model'],
        best_model_name,
        X_train,
        X_test,
        feature_names
    )

    # Also analyze k-NN for comparison
    if 'k-NN' in results:
        print("\nAnalyzing k-NN model for comparison")
        perform_shap_analysis(
            results['k-NN']['model'],
            'k-NN',
            X_train,
            X_test,
            feature_names
        )

    print("\n=== All operations completed successfully ===")

    return results, kfold_results, X_test, y_test

if __name__ == "__main__":
    main()
```
